```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_modal import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
from statsmodels.tsa.seasonal import seasonal_decompose
from stats.models.tsa.stattools import coint
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import warnings
```

```python
import json
from IPython import get_ipython

file_path = '/content/sample_data/AB_Testing.ipynb'

try:
    with open(file_path, 'r', encoding='utf-8') as f:
        nb_content = json.load(f)

    code_cells = [cell for cell in nb_content.get('cells', []) if cell.get('cell_type') == 'code']

    print(f"Found {len(code_cells)} code cells in the notebook. Executing them with IPython run_cell:\n")

    ipython = get_ipython()
    for idx, cell in enumerate(code_cells):
        source_code = "".join(cell.get('source', []))
        print(f"--- Executing Cell {idx+1} ---")
        # Use run_cell to handle magics like %matplotlib inline
        res = ipython.run_cell(source_code)
        if not res.success:
            print(f"Cell {idx+1} execution failed.")
            if res.error_before_exec:
                print(res.error_before_exec)
            if res.error_in_exec:
                print(res.error_in_exec)
            break
        print("-" * 40)

except Exception as e:
    print(f"An error occurred during file processing: {e}")
```
